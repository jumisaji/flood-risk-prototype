# Demo — API vs notebook parity check

Runs the same four features through the three tabular models locally and averages
them (equal-weight soft-vote), exactly like the deployed `/predict` endpoint does
with the default ensemble. The number this prints should match the API response.



```python
import joblib
import pandas as pd

FEATURES = {
    "level_lag1": 0.73,
    "level_lag2": 0.69
    "level_roll7": 0.68,
    "level_change3": 0.06,
}

row = pd.DataFrame([FEATURES])

lr  = joblib.load("../backend/models/logistic_regression_real.joblib")
rf  = joblib.load("../backend/models/random_forest.joblib")
xgb = joblib.load("../backend/models/xgboost.joblib")

probs = {
    "Logistic Regression": float(lr.predict_proba(row)[0][1]),
    "Random Forest":       float(rf.predict_proba(row)[0][1]),
    "XGBoost":              float(xgb.predict_proba(row)[0][1]),
}

ensemble = sum(probs.values()) / len(probs)

print(f"Features: {FEATURES}")
print("-" * 48)
for name, p in probs.items():
    print(f"  {name:22s} {p:.4f}")
print("-" * 48)
print(f"Ensemble (mean of the three): {round(ensemble, 4)}")
print("\nCompare this to the API response for POST /predict with the same features.")
```
